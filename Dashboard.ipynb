````python
import pandas as pd  #  analizar, limpiar y manipular datos
import matplotlib.pyplot as plt  # crear visualizaciones de datos estáticas, animadas e interactivas.

# Cargar archivo
df = pd.read_excel(
    "/content/sample_data/mdi_personasdesaparecidas_pm_2026_enero_agosto.xlsx",
    sheet_name="1. Pdesaparecidas"
)

# Contar desaparecidos por provincia
provincia_counts = df["provincia"].value_counts()

# Tomar las 10 provincias con más casos
top10 = provincia_counts.head(10)

# Agrupar el resto
otras = provincia_counts.iloc[10:].sum()

# Crear serie para el gráfico
datos_grafico = pd.concat([
    top10,
    pd.Series({"OTRAS": otras})
])

# Gráfico circular
plt.figure(figsize=(10,10))
plt.pie(
    # Datos numéricos que se representarán en cada sector
    datos_grafico,
    # Etiquetas que aparecerán junto a cada sector
    labels=datos_grafico.index,
     # Mostrar el porcentaje de cada sector con 1 decimal
    autopct="%1.1f%%",
     # Rotar el gráfico para que comience en 90 grados
    startangle=90
)

plt.title("Personas desaparecidas por provincia\n(Enero - Agosto 2026)")
plt.tight_layout()
plt.show()
````

````python
# ==========================================================
# INFORME GRÁFICO DE BARRAS
# PERSONAS DESAPARECIDAS POR RANGO DE EDAD
# ==========================================================

# Importar librerías
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------------------------------------
# 1. Cargar archivo Excel
# ----------------------------------------------------------

archivo = "mdi_personasdesaparecidas_pm_2026_enero_agosto.xlsx"

df = pd.read_excel(
    archivo,
    sheet_name="1. Pdesaparecidas"
)

# ----------------------------------------------------------
# 2. Mostrar columnas disponibles
# ----------------------------------------------------------

print("Columnas encontradas:")
print(df.columns.tolist())

# ----------------------------------------------------------
# 3. Contar registros por rango de edad
# ----------------------------------------------------------

conteo_rango = df["rango_edad"].value_counts()

# Ordenar de mayor a menor
conteo_rango = conteo_rango.sort_values(ascending=False)

print("\nDesaparecidos por rango de edad:")
print(conteo_rango)

# ----------------------------------------------------------
# 4. Calcular porcentajes
# ----------------------------------------------------------

porcentajes = round(
    (conteo_rango / conteo_rango.sum()) * 100,
    2
)

# ----------------------------------------------------------
# 5. Crear tabla resumen
# ----------------------------------------------------------

resumen = pd.DataFrame({
    "Rango de Edad": conteo_rango.index,
    "Cantidad": conteo_rango.values,
    "Porcentaje (%)": porcentajes.values
})

print("\nResumen estadístico")
print(resumen)

# ----------------------------------------------------------
# 6. Crear gráfico de barras
# ----------------------------------------------------------

plt.figure(figsize=(12, 6))

barras = plt.bar(
    conteo_rango.index,
    conteo_rango.values
)

# ----------------------------------------------------------
# 7. Etiquetas sobre cada barra
# ----------------------------------------------------------

for barra, cantidad, porcentaje in zip(
        barras,
        conteo_rango.values,
        porcentajes.values):

    plt.text(
        barra.get_x() + barra.get_width()/2,
        cantidad,
        f"{cantidad}\n({porcentaje}%)",
        ha='center',
        va='bottom'
    )

# ----------------------------------------------------------
# 8. Personalización del gráfico
# ----------------------------------------------------------

plt.title(
    "Personas Desaparecidas por Rango de Edad\nEnero - Agosto 2026",
    fontsize=14
)

plt.xlabel("Rango de Edad")
plt.ylabel("Número de Personas")

plt.xticks(rotation=45)

plt.grid(
    axis='y',
    linestyle='--',
    alpha=0.5
)

plt.tight_layout()

# ----------------------------------------------------------
# 9. Guardar gráfico
# ----------------------------------------------------------

plt.savefig(
    "desaparecidos_rango_edad.png",
    dpi=300,
    bbox_inches="tight"
)

# ----------------------------------------------------------
# 10. Mostrar gráfico
# ----------------------------------------------------------

plt.show()
````

````python
####Correcciones de la base de datos una vez identificado lo anterior E-Gracia###
import numpy as np

print("Iniciando proceso de limpieza (EDA)...")

# 1. Tratamiento de valores "NO_APLICA"
# Reemplazamos el texto por nulos matemáticos (NaN) para que no dañen los promedios
df.replace('NO_APLICA', np.nan, inplace=True)

# 2. Corrección de Coordenadas para el Mapa
# Forzamos a texto, cambiamos la coma por punto y convertimos a decimal (float)
if 'latitud' in df.columns and 'longitud' in df.columns:
    df['latitud'] = df['latitud'].astype(str).str.replace(',', '.').astype(float)
    df['longitud'] = df['longitud'].astype(str).str.replace(',', '.').astype(float)

# 3. Conversión de Fechas
# Aunque Excel a veces pre-formatea las fechas, forzamos el formato datetime de Pandas
columnas_fecha = ['fecha_desaparicion', 'fecha_denuncia', 'fecha_conocimiento', 'fecha_localizacion']
for col in columnas_fecha:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')

# 4. Creación de nueva variable estratégica (Dato de negocio para el Dashboard)
if 'fecha_denuncia' in df.columns and 'fecha_desaparicion' in df.columns:
    df['dias_demora_denuncia'] = (df['fecha_denuncia'] - df['fecha_desaparicion']).dt.days

# Exportamos el archivo limpio para que el Integrante 3 lo use en Streamlit
df.to_csv('desaparecidos_limpio.csv', index=False)

print("¡Limpieza terminada con éxito!")
print("Archivo 'desaparecidos_limpio.csv' generado para conectar con el Dashboard.")
````

````python

# ==========================================================
# DASHBOARD AVANZADO
# PERSONAS DESAPARECIDAS
# ==========================================================

# ----------------------------------------------------------
# 1. IMPORTAR LIBRERÍAS
# ----------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------------------------------------
# 2. CARGAR ARCHIVO
# ----------------------------------------------------------

archivo = "desaparecidos_limpio.csv"

df = pd.read_csv(
    archivo,
    encoding="utf-8"
)

# ----------------------------------------------------------
# 3. CONVERTIR FECHAS
# ----------------------------------------------------------

df["fecha_desaparicion"] = pd.to_datetime(
    df["fecha_desaparicion"],
    errors="coerce"
)

df["fecha_denuncia"] = pd.to_datetime(
    df["fecha_denuncia"],
    errors="coerce"
)

df["fecha_localizacion"] = pd.to_datetime(
    df["fecha_localizacion"],
    errors="coerce"
)

# ----------------------------------------------------------
# 4. CALCULAR DÍAS HASTA LOCALIZACIÓN
# ----------------------------------------------------------

df["dias_hasta_localizacion"] = (
    df["fecha_localizacion"]
    - df["fecha_desaparicion"]
).dt.days

# ----------------------------------------------------------
# 5. CONFIGURACIÓN DEL DASHBOARD
# ----------------------------------------------------------

fig = plt.figure(figsize=(20, 14))

# ==========================================================
# GRÁFICO 1
# POLÍGONO DE FRECUENCIA
# DESAPARICIONES POR MES
# ==========================================================

ax1 = plt.subplot(2, 3, 1)

meses = (
    df["fecha_desaparicion"]
    .dt.month
    .value_counts()
    .sort_index()
)

nombres_meses = {
    1:"Ene",
    2:"Feb",
    3:"Mar",
    4:"Abr",
    5:"May",
    6:"Jun",
    7:"Jul",
    8:"Ago",
    9:"Sep",
    10:"Oct",
    11:"Nov",
    12:"Dic"
}

etiquetas = [
    nombres_meses[m]
    for m in meses.index
]

ax1.plot(
    etiquetas,
    meses.values,
    marker="o",
    linewidth=2
)

ax1.set_title(
    "Desapariciones por Mes de Ocurrencia"
)

ax1.set_xlabel("Mes")
ax1.set_ylabel("Casos")
ax1.grid(True)

# ==========================================================
# GRÁFICO 2
# CASOS POR RANGO DE EDAD Y SEXO
# ==========================================================

ax2 = plt.subplot(2, 3, 2)

tabla_edad = pd.crosstab(
    df["rango_edad"],
    df["sexo"]
)

tabla_edad.plot(
    kind="bar",
    ax=ax2
)

ax2.set_title(
    "Casos por Grupo de Edad y Sexo"
)

ax2.set_xlabel("Grupo de Edad")
ax2.set_ylabel("Casos")
ax2.tick_params(axis='x', rotation=45)

# ==========================================================
# GRÁFICO 3
# % SIN LOCALIZAR POR PROVINCIA
# ==========================================================

ax3 = plt.subplot(2, 3, 3)

resumen = (
    df.groupby("provincia")
    .agg(
        total=("provincia", "size"),
        desaparecidos=(
            "situacion_actual",
            lambda x:
            (x == "DESAPARECIDO").sum()
        )
    )
)

resumen = resumen[
    resumen["total"] >= 100
]

resumen["porcentaje"] = (
    resumen["desaparecidos"]
    / resumen["total"]
) * 100

resumen = resumen.sort_values(
    "porcentaje",
    ascending=True
)

ax3.barh(
    resumen.index,
    resumen["porcentaje"]
)

ax3.set_title(
    "% Casos sin Localizar por Provincia"
)

ax3.set_xlabel("Porcentaje")

# ==========================================================
# GRÁFICO 4
# DEMORA EN DENUNCIAR
# ==========================================================

ax4 = plt.subplot(2, 3, 4)

demora = pd.to_numeric(
    df["dias_demora_denuncia"],
    errors="coerce"
)

ax4.hist(
    demora.dropna(),
    bins=20
)

ax4.set_title(
    "Demora en Denunciar"
)

ax4.set_xlabel("Días")
ax4.set_ylabel("Frecuencia")

# ==========================================================
# GRÁFICO 5
# TIEMPO HASTA LA LOCALIZACIÓN
# ==========================================================

ax5 = plt.subplot(2, 3, 5)

localizados = df[
    df["dias_hasta_localizacion"].notna()
]

ax5.hist(
    localizados["dias_hasta_localizacion"],
    bins=20
)

ax5.set_title(
    "Tiempo hasta la Localización"
)

ax5.set_xlabel("Días")
ax5.set_ylabel("Frecuencia")

# ==========================================================
# GRÁFICO 6
# TOP 10 PROVINCIAS
# ==========================================================

ax6 = plt.subplot(2, 3, 6)

top_prov = (
    df["provincia"]
    .value_counts()
    .head(10)
    .sort_values()
)

ax6.barh(
    top_prov.index,
    top_prov.values
)

ax6.set_title(
    "Top 10 Provincias con Más Casos"
)

ax6.set_xlabel("Número de Casos")

# ----------------------------------------------------------
# AJUSTAR DISEÑO
# ----------------------------------------------------------

plt.tight_layout()

# ----------------------------------------------------------
# GUARDAR DASHBOARD
# ----------------------------------------------------------

plt.savefig(
    "dashboard_desaparecidos_avanzado.png",
    dpi=300,
    bbox_inches="tight"
)

# ----------------------------------------------------------
# MOSTRAR DASHBOARD
# ----------------------------------------------------------

plt.show()
```
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- Cargar el archivo de datos limpio ---
# Se espera que el archivo 'desaparecidos_limpio.csv' esté disponible,
# el cual se generó en pasos anteriores tras la limpieza de datos.
archivo = "/content/sample_data/desaparecidos_limpio.csv"
df = pd.read_csv(archivo, encoding="utf-8")

# --- Conversión de columnas de fecha a formato datetime ---
# Esto es crucial para realizar operaciones basadas en tiempo, como extraer el mes o calcular diferencias de días.
df["fecha_desaparicion"] = pd.to_datetime(df["fecha_desaparicion"], errors="coerce")
df["fecha_denuncia"] = pd.to_datetime(df["fecha_denuncia"], errors="coerce")
df["fecha_localizacion"] = pd.to_datetime(df["fecha_localizacion"], errors="coerce")

# --- Cálculo de días hasta la localización ---
# Esta nueva columna indica el tiempo transcurrido desde la desaparición hasta la localización.
df["dias_hasta_localizacion"] = (df["fecha_localizacion"] - df["fecha_desaparicion"]).dt.days

# --- Configuración general del dashboard ---
# Se crea una figura con un tamaño específico para contener múltiples gráficos (2 filas, 3 columnas).
fig = plt.figure(figsize=(20, 14))

# ==========================================================
# GRÁFICO 1: POLÍGONO DE FRECUENCIA - DESAPARICIONES POR MES
# ==========================================================
# Este gráfico muestra la distribución de las desapariciones a lo largo de los meses.
ax1 = plt.subplot(2, 3, 1)

# Contar las desapariciones por mes y ordenar por índice (mes).
meses = df["fecha_desaparicion"].dt.month.value_counts().sort_index()

# Mapear números de mes a nombres abreviados para una mejor visualización.
nombres_meses = {
    1:"Ene", 2:"Feb", 3:"Mar", 4:"Abr", 5:"May", 6:"Jun",
    7:"Jul", 8:"Ago", 9:"Sep", 10:"Oct", 11:"Nov", 12:"Dic"
}

# Generar etiquetas de mes para el eje X.
etiquetas = [nombres_meses[m] for m in meses.index]

# Dibujar el polígono de frecuencia.
ax1.plot(etiquetas, meses.values, marker="o", linewidth=2)
ax1.set_title("Desapariciones por Mes de Ocurrencia")
ax1.set_xlabel("Mes")
ax1.set_ylabel("Casos")
ax1.grid(True)

# ==========================================================
# GRÁFICO 2: CASOS POR GRUPO DE EDAD Y SEXO
# ==========================================================
# Este gráfico de barras apiladas (o agrupadas) muestra la cantidad de casos
# desagregados por rango de edad y por sexo (mujeres y hombres).
ax2 = plt.subplot(2, 3, 2)

# Crear una tabla de contingencia para contar casos por rango de edad y sexo.
tabla_edad = pd.crosstab(df["rango_edad"], df["sexo"])

# Generar el gráfico de barras.
tabla_edad.plot(kind="bar", ax=ax2)
ax2.set_title("Casos por Grupo de Edad y Sexo")
ax2.set_xlabel("Grupo de Edad")
ax2.set_ylabel("Casos")
ax2.tick_params(axis='x', rotation=45) # Rotar etiquetas para mejor legibilidad.

# ==========================================================
# GRÁFICO 3: PORCENTAJE DE CASOS SIN LOCALIZAR POR PROVINCIA
# ==========================================================
# Muestra el porcentaje de casos aún "DESAPARECIDO" por provincia,
# considerando solo provincias con al menos 100 casos totales.
ax3 = plt.subplot(2, 3, 3)

# Agrupar por provincia y calcular el total de casos y los casos 'DESAPARECIDO'.
resumen = df.groupby("provincia").agg(
    total=("provincia", "size"),
    desaparecidos=("situacion_actual", lambda x: (x == "DESAPARECIDO").sum())
)

# Filtrar provincias con al menos 100 casos totales.
resumen = resumen[resumen["total"] >= 100]

# Calcular el porcentaje de casos sin localizar.
resumen["porcentaje"] = (resumen["desaparecidos"] / resumen["total"]) * 100

# Ordenar por porcentaje para una visualización más clara.
resumen = resumen.sort_values("porcentaje", ascending=True)

# Generar el gráfico de barras horizontales.
ax3.barh(resumen.index, resumen["porcentaje"])
ax3.set_title("% Casos sin Localizar por Provincia")
ax3.set_xlabel("Porcentaje")

# ==========================================================
# GRÁFICO 4: DEMORA EN DENUNCIAR
# ==========================================================
# Histograma que muestra la distribución de los días de demora entre la desaparición y la denuncia.
ax4 = plt.subplot(2, 3, 4)

# Asegurar que la columna 'dias_demora_denuncia' sea numérica, manejando posibles errores.
demora = pd.to_numeric(df["dias_demora_denuncia"], errors="coerce")

# Dibujar el histograma, eliminando valores nulos para evitar errores.
ax4.hist(demora.dropna(), bins=20) # 20 "bins" o intervalos para la distribución.
ax4.set_title("Demora en Denunciar")
ax4.set_xlabel("Días")
ax4.set_ylabel("Frecuencia")

# ==========================================================
# GRÁFICO 5: TIEMPO HASTA LA LOCALIZACIÓN
# ==========================================================
# Histograma que muestra la distribución de los días que tardó la localización de las personas.
ax5 = plt.subplot(2, 3, 5)

# Filtrar solo los casos que tienen una fecha de localización (es decir, fueron localizados).
localizados = df[df["dias_hasta_localizacion"].notna()]

# Dibujar el histograma para el tiempo hasta la localización.
ax5.hist(localizados["dias_hasta_localizacion"], bins=20)
ax5.set_title("Tiempo hasta la Localización")
ax5.set_xlabel("Días")
ax5.set_ylabel("Frecuencia")

# ==========================================================
# GRÁFICO 6: TOP 10 PROVINCIAS CON MÁS CASOS
# ==========================================================
# Este gráfico muestra las diez provincias con la mayor cantidad de casos de desaparición.
ax6 = plt.subplot(2, 3, 6)

# Contar los casos por provincia, tomar el top 10 y ordenarlos.
top_prov = df["provincia"].value_counts().head(10).sort_values()

# Generar el gráfico de barras horizontales.
ax6.barh(top_prov.index, top_prov.values)
ax6.set_title("Top 10 Provincias con Más Casos")
ax6.set_xlabel("Número de Casos")

# --- Ajustar el diseño de los gráficos ---
# Esto evita que los títulos y etiquetas se superpongan, haciendo el dashboard más legible.
plt.tight_layout()

# --- Guardar el dashboard como una imagen ---
# El dashboard completo se guarda en un archivo PNG con alta resolución.
plt.savefig("dashboard_desaparecidos_avanzado.png", dpi=300, bbox_inches="tight")

# --- Mostrar el dashboard ---
# Finalmente, se muestra el dashboard con todos los gráficos.
plt.show()
````
